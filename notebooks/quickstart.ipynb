```python
# Cell 1: Environment Setup & Cloning the Repository
!apt-get update -qq && apt-get install -y php8.2-cli php8.2-curl php8.2-mbstring git
!git clone https://github.com/eidcloud/eidcloud-api-cli.git /content/eidcloud-api-cli
%cd /content/eidcloud-api-cli
!php -v
!php tests/run_tests.php
```

```python
# Cell 2: Convert OpenAPI Specification into Interactive CLI Executable
!php bin/eidcloud-api-cli generate examples/sample-openapi.json --name=sample-cli --out=/content/bin
!chmod +x /content/bin/sample-cli
!php /content/bin/sample-cli --help
```

```python
# Cell 3: Execute Direct Endpoint Calls with ANSI Table and CSV Rendering
print("--- 1. Direct Table Output ---")
!php bin/eidcloud-api-cli call examples/sample-openapi.json /users GET --limit=5 --table

print("\n--- 2. Direct CSV Output ---")
!php bin/eidcloud-api-cli call examples/sample-openapi.json /posts GET --limit=3 --csv

print("\n--- 3. Using the Generated CLI ---")
!php /content/bin/sample-cli users list --limit=3 --table
```
